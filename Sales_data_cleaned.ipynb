```python
#Standardise columns name
df.columns=(
    df.columns
    .str.replace(' ','_')
    .str.lower()
    .str.strip()
)
```

```python
df.describe(include="all")
```

```python
missing=df.isnull().mean()*100
print(missing)
```

```python
n_dupes=df.duplicated().sum()
n_dupes
```

```python
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df["sales"].dtype
```

```python
#Convertir sales
df["sales"] = pd.to_numeric(
    df["sales"].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
)
```

```python
print(df["sales"].dtype)
print(df["sales"].isnull().sum())
```

```python
#Convertir les dates
df["order_date"] = pd.to_datetime(
    df["order_date"],
    dayfirst=True,
    errors="coerce"
)
```

```python
df["ship_date"] = pd.to_datetime(
    df["ship_date"],
    dayfirst=True,
    errors="coerce")
```

```python
print(df["order_date"].dtype)
print(df["ship_date"].dtype)
```

```python
print("Dates commande invalides :", df["order_date"].isnull().sum())
print("Dates expédition invalides :", df["ship_date"].isnull().sum())
```

```python
df["year"].dtype
```

```python
#la cohérence entre year et order_date
year_from_date = df["order_date"].dt.year

print(
    "Nombre d'incohérences :",
    (year_from_date != df["year"]).sum()
)
```

```python
print(df["quantity"].describe())
```

```python
print("Quantités <= 0 :", (df["quantity"] <= 0).sum())
```

```python
#Vérifier les remises
df["discount"]
```

```python
print(
    "Discount < 0 :",
    (df["discount"] < 0).sum()
)

print(
    "Discount > 1 :",
    (df["discount"] > 1).sum()
)
```

```python
#Vérifier le profit
print(df["profit"].describe())
```

```python
print("Profit négatif :", (df["profit"] < 0).sum())
```

```python
#les frais de livraison
print(df["shipping_cost"].describe())
```

```python
print(
    "Shipping cost négatif :",
    (df["shipping_cost"] < 0).sum()
)
```

```python
#Vérifier la logique des dates
df["shipping_days"] = (
    df["ship_date"] - df["order_date"]
).dt.days
```

```python
df["shipping_days"].describe()
```

```python
print(
    "Dates incohérentes :",
    (df["shipping_days"] < 0).sum()
)
```

Créer les variables utiles pour l'analyse

Maintenant que les données sont propres, nous pouvons créer des variables analytiques.

```python
df["month"] = df["order_date"].dt.month
```

```python
df["month_name"] = df["order_date"].dt.month_name()
```

```python
df["quarter"] = df["order_date"].dt.quarter
```

```python
df["day"] = df["order_date"].dt.day
```

```python
df["day_of_week"] = df["order_date"].dt.day_name()
```

```python
#creer de la marge
df["profit_margin"] = np.where(
    df["sales"] != 0,
    (df["profit"] / df["sales"]) * 100,
    0
)
```

```python
print(
    "Sales = 0 :",
    (df["sales"] == 0).sum()
)
```
